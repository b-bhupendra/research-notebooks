# Exploratory Data Analysis & Statistical Signal Detection

This notebook demonstrates automated distribution profiling, correlation extraction, and multivariate anomaly detection.

```python
import numpy as np
import pandas as pd

# Synthesizing benchmark dataset
np.random.seed(42)
n_samples = 1000

data = pd.DataFrame({
    'feature_latency_ms': np.random.exponential(scale=25.0, size=n_samples),
    'feature_throughput_rps': np.random.normal(loc=1200, scale=150, size=n_samples),
    'feature_error_rate': np.random.beta(a=1, b=50, size=n_samples)
})

data.describe()
```

## Correlation & Feature Relationships

Evaluating pairwise correlation matrix across system performance features.

```python
correlation_matrix = data.corr()
print("Correlation Matrix:")
print(correlation_matrix)
```
